# 01 — Setup check

What this notebook does: checks that Python, PyTorch, and Avalanche are installed and working, and reports whether a GPU is available (and which one). It doesn't load any data yet — it's just confirming the environment is ready before we build anything on top of it.

Run this cell by cell. If a cell prints a clear error with an install hint, follow the hint, re-run that cell, then continue.

```python
import sys
print(f"Python version: {sys.version}")
```

## Check PyTorch and GPU access

This checks whether PyTorch is installed and whether it can see a GPU — either the local GPU on this machine, or the GPU on a RunPod instance if that's where this notebook is running.

```python
try:
    import torch
    print(f"torch version: {torch.__version__}")
    gpu_available = torch.cuda.is_available()
    print(f"GPU available: {gpu_available}")
    if gpu_available:
        print(f"GPU name: {torch.cuda.get_device_name(0)}")
        print(f"GPU memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
    else:
        print("No GPU detected — this will be slow for anything beyond a quick test.")
except ImportError:
    print("torch is not installed. Install it with: pip install torch torchvision")
    print("(On a machine with an NVIDIA GPU, follow the command for your CUDA version from https://pytorch.org/get-started/locally/ instead of a plain pip install, so it installs the GPU-enabled build.)")
```

## Confirm the GPU actually runs a computation

Just detecting a GPU isn't proof it works — this runs a tiny matrix multiplication on it to make sure.

```python
if 'torch' in dir() and torch.cuda.is_available():
    a = torch.rand(1000, 1000, device="cuda")
    b = torch.rand(1000, 1000, device="cuda")
    c = a @ b
    torch.cuda.synchronize()
    print(f"GPU matrix multiply worked. Result shape: {c.shape}, sample value: {c[0,0].item():.3f}")
else:
    print("Skipping GPU computation check — no GPU available.")
```
